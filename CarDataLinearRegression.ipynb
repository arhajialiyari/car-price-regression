```python
# Compute the correlation matrix
corr = df2.corr(numeric_only=True)

# Create the heatmap without annotations
plt.figure(figsize=(8, 6))  # Increase figure size
sns.heatmap(corr, linewidths=.5,
            cmap="Oranges", mask=np.triu(corr, k=1), vmax=1, vmin=-1)

# Manually add annotations
for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):
        if i >= j:
            plt.text(j + 0.5, i + 0.5, f"{corr.iloc[i, j]:.2f}",
                     ha='center', va='center', color='black')

plt.show()
```

```python
# Function to create subplots
def create_subplots(hue, position, title, fig, gs):
    ax_main = fig.add_subplot(gs[position[0]:position[1], position[2]:position[3]])
    sns.scatterplot(data=df2, x='Present_Price', y='Selling_Price', hue=hue, ax=ax_main)
    sns.regplot(data=df2, x='Present_Price', y='Selling_Price', scatter=False, ax=ax_main)
    sns.rugplot(data=df2, x='Present_Price', y='Selling_Price', ax=ax_main)
    ax_main.set_title(title)
    
    # Density plot for x-axis (Present_Price)
    ax_x = fig.add_subplot(gs[position[4], position[2]:position[3]], sharex=ax_main)
    sns.histplot(data=df2, x='Present_Price', hue=hue, kde=True, element='step', fill=True, ax=ax_x)
    ax_x.axis('off')
```

```python
# Create a figure with subplots
fig = plt.figure(figsize=(12, 12))
gs = GridSpec(8, 8, figure=fig, hspace=0.4, wspace=0.8)

# Define the hues and positions for subplots
hues = ['Fuel_Type', 'Seller_Type']
positions = [
    (1, 4, 0, 4, 0),  # for Fuel_Type
    (1, 4, 4, 8, 0),  # for Seller_Type
]

# Loop through hues and create subplots
for hue, pos in zip(hues, positions):
    create_subplots(hue, pos, f'Selling Price vs Present Price ({hue})', fig, gs)

plt.tight_layout()
plt.show()
```

```python
df3 = df2.copy()
max_year = df2['Year'].max()
df3.insert(0, 'Age', max_year - df3['Year'] + 1)
df3.drop('Year', axis=1, inplace=True)
df3.sample(5)
```

```python
df4 = df3.copy()
df4.replace({'Fuel_Type': {'Petrol':2,'Diesel':3,'CNG':4}} , inplace=True)
df4.replace({'Seller_Type': {'Dealer':2,'Individual':3}} , inplace=True)
df4.replace({'Transmission': {'Manual':2,'Automatic':3}} , inplace=True)
df4.sample(5)
```

```python
# Compute the correlation matrix
corr = df4.corr(numeric_only=True)

# Create the heatmap without annotations
plt.figure(figsize=(8, 6))  # Increase figure size
sns.heatmap(corr, linewidths=.5,
            cmap="Oranges", mask=np.triu(corr, k=1), vmax=1, vmin=-1)

# Manually add annotations
for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):
        if i >= j:
            plt.text(j + 0.5, i + 0.5, f"{corr.iloc[i, j]:.2f}",
                     ha='center', va='center', color='black')

plt.show()
```

```python
def create_weights_table(model, feature_names):

    weights = np.append(model.intercept_, model.coef_)
    columns = ['intercept'] + feature_names

    weights_df = pd.DataFrame({
        'Parameters': ['b'] + [f'w_{i+1}' for i in range(len(feature_names))],
        'Columns': columns,
        'weights': weights.flatten()
    })

    return weights_df
```

```python
def show_results(learn_model, test_size, X_train, X_test, y_train, y_real, y_pred, y_pred_train, feature_names, model):
    print(learn_model)
    print('_' * 30)
    
    print(f"\033[1mtest size = {test_size}:\033[0m")

    print(colored('='*30, 'blue'), colored('Shape', 'blue'), colored('='*30, 'blue'))
    
    print(colored("X train{:>1}: {}".format("", X_train.shape), 'blue'))
    print(colored("X test{:>2}: {}".format("", X_test.shape), 'blue'))
    print(colored("y train{:>1}: {}".format("", y_train.shape), 'blue'))
    print(colored("y test{:>2}: {}".format("", y_real.shape), 'blue'))

    print(colored('='*30, 'green'), colored(' Evaluation Metrics ', 'green'), colored('='*30, 'green'))

    print(colored("Mean Absolute Error{:>5}: {:.4f}".format("", metrics.mean_absolute_error(y_real, y_pred)), 'green'))
    print(colored("Mean Squared Error{:>6}: {:.4f}".format("", metrics.mean_squared_error(y_real, y_pred)), 'green'))
    print(colored("Root Mean Squared Error{:>1}: {:.4f}".format("", np.sqrt(metrics.mean_squared_error(y_real, y_pred))), 'green'))
    print(colored("R2Score{:>17}: {:.4f}".format("", metrics.r2_score(y_real, y_pred)), 'green'))
    print(colored("R2Score Train{:>11}: {:.4f}".format("", metrics.r2_score(y_train, y_pred_train)), 'green'))

    weights_df = create_weights_table(model, feature_names)
    display(weights_df)
```

```python
def scale_data(df, target_column):
    scaler = preprocessing.MinMaxScaler(feature_range=(0, 1))
    scaled_data = scaler.fit_transform(df.drop(columns=[target_column]))
    scaled_df = pd.DataFrame(scaled_data, columns=list(df.drop(target_column, axis=1).columns))
    scaled_df[target_column] = df[target_column]
    return scaled_df
```

```python
def create_model(df, learn_model, target_column, test_size=0.2, random_state=42, show_result=True):
    
    scaled_df = scale_data(df, target_column)

    X, y = scaled_df.drop(columns=target_column), df[target_column].values.reshape(-1, 1)
    X_train, X_test, y_train, y_real = train_test_split(X, y, test_size=test_size, random_state=random_state)

    model = learn_model().fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_pred_train = model.predict(X_train)

    feature_names = X_train.columns.tolist()

    if show_result:
        show_results(learn_model, test_size, X_train, X_test, y_train, y_real, y_pred, y_pred_train, feature_names, model)

    return X_train, X_test, y_train, y_real, y_pred, y_pred_train, model, feature_names
```

```python
def visualize_model(X_train, X_test, y_train, y_pred, feature_names):
    columns = feature_names 

    for col in columns:
        fig, ax = plt.subplots(1, 3, figsize=(15, 4))

        # Plot for training data
        ax[0].scatter(X_train[col], y_train, label='real', color='orange')
        ax[0].set_title('Training Data', fontsize=14, fontweight='bold')
        ax[0].set_xlabel(col)
        ax[0].set_ylabel('Selling_Price')

        # Plot for predictions
        ax[1].scatter(X_test[col], y_pred, label='predict', color='blue')
        ax[1].set_title('Prediction', fontsize=14, fontweight='bold')
        ax[1].set_xlabel(col)
        ax[1].set_ylabel('Selling_Price')

        # Combined plot for training data vs predictions
        ax[2].scatter(X_train[col], y_train, label='real', color='orange')
        ax[2].scatter(X_test[col], y_pred, label='predict', color='blue', alpha=0.6)
        ax[2].set_title('Training Data vs Prediction', fontsize=14, fontweight='bold')
        ax[2].set_xlabel(col)
        ax[2].set_ylabel('Selling_Price')

        plt.legend()
        plt.show()
```

```python
def compare_real_vs_predicted(x_test, y_real, y_pred, columns):
    x_test.insert(0, 'y_real', y_real.flatten())
    x_test.insert(0, 'y_pred', y_pred.flatten())

    for col in columns:
        new_df = x_test.sort_values(by=[col])
        plt.scatter(new_df[col], new_df.y_real, marker='.', color='black', label='real')
        plt.plot(new_df[col], new_df.y_pred, color='r', alpha=0.6, label='model')
        plt.title('Car Price Prediction Model', fontsize=14, fontweight='bold')
        plt.xlabel(col)
        plt.ylabel('Selling_Price')
        plt.legend()
        plt.show()
```
